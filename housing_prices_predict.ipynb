```python
#import libraries
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv(r'C:\Users\FPTSHOP AI NGHIA\PycharmProjects\housing_prices_prediction\data\train.csv', index_col='Id')
data.head(5)
```

```python
data.columns
```

#feature

```python
features= ["LotArea", "YearBuilt", "1stFlrSF", "2ndFlrSF", "FullBath", "BedroomAbvGr", "TotRmsAbvGrd"]
```

```python
X = data[features]
y = data["SalePrice"]
y.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_valid, y_train, y_valid = train_test_split(X,y,train_size= 0.8, test_size=0.2,random_state=0)
```

```python
X_train.shape
```

```python
X_valid.shape
```

```python
from sklearn.tree import DecisionTreeRegressor
dt_model = DecisionTreeRegressor(random_state=1)
dt_model.fit(X_train,y_train)
y_predict = dt_model.predict(X_valid.head(5))
```

```python
pd.DataFrame({'y' : y_valid.head(5), 'y_predict' : y_predict})
```

```python
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
rf = RandomForestRegressor(random_state=1)
rf.fit(X_train,y_train)
rf_val_predict = rf.predict(X_valid)
rf_val_predict[:5]
pd.DataFrame({'y': y_valid.head(5), 'y_predict' : rf_val_predict[:5]})
```
